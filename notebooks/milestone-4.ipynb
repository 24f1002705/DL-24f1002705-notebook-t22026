## MAP@3 function

```python
# takes two inputs, one is a list of all correct answers, the other a list of corresponding predictions
def map_at_3(y_true, y_pred):
    score = 0.0

    if len(y_true) != len(y_pred):
        raise ValueError("Length mismatch")

    for i in range(len(y_true)):
        pred = y_pred[i]
        true = y_true[i]
        if true in pred:
            for j in range(len(pred)):
                if pred[j] == true:
                    score += 1/(j+1)
                    break
                
    return score/len(y_true)
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print(device)
print(torch.cuda.get_device_name(0))
```

```python
# Train/Validation split
train_df, val_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df['answer'],
    random_state=42
)

print(len(train_df))
print(len(val_df))
```

```python
def expand_dataset(df):
    samples = []

    for _, row in df.iterrows():
        
        for option in ["A","B","C","D","E"]:
            samples.append({
                'prompt': row['prompt'],

                'option': row[option],
                
                "label": 1 if option == row["answer"] else 0,
    
                "question_id": row["id"]
            })

    return pd.DataFrame(samples)
```

```python
train_pairs = expand_dataset(train_df)

val_pairs = expand_dataset(val_df)
```

```python
train_pairs.head()
```

```python
# Load Deberta 
MODEL_NAME = "microsoft/deberta-v3-base"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)
```

```python
# create MCQDATASET class
class MCQDataset(Dataset):
    def __init__(self, dataframe):
        self.prompts = dataframe['prompt'].tolist()
        self.options = dataframe['option'].tolist()
        self.labels = dataframe['label'].tolist()

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        encoding = tokenizer(
            self.prompts[idx],
            self.options[idx],
            truncation=True,
            max_length=256
        )

        return {
            "input_ids":
                encoding['input_ids'],

            "attention_mask":
                encoding['attention_mask'],

            "labels": int(self.labels[idx])
        }
```

```python
train_dataset = MCQDataset(train_pairs)
val_dataset = MCQDataset(val_pairs)
```

```python
# data collator
data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer,
    return_tensors="pt"
)
```

```python
train_loader =  DataLoader(
    train_dataset,
    batch_size = 16,
    shuffle=True,
    collate_fn=data_collator
)

val_loader = DataLoader(
    val_dataset,
    batch_size= 16,
    shuffle=False,
    collate_fn=data_collator
)
```

## Training Loop

### Start wandb run

```python
import wandb

wandb.init(

    project="24f1002705-DLGenAI-t12026",

    name="deberta-v3-base",

    config={

        "model": MODEL_NAME,
        "epochs": epochs,
        "batch_size": 16,
        "learning_rate": 2e-5,
        "max_length": 256,
        "optimizer": "AdamW",
        "scheduler": "Linear Warmup"

    }

)
```

```python
lowest_val_loss = 0

for epoch in range(epochs):

    ##################################
    # Training
    ##################################

    model.train()

    train_loss = 0

    for batch in train_loader:

        input_ids = batch["input_ids"].to(device)

        attention_mask = batch["attention_mask"].to(device)

        labels = batch["labels"].long().to(device)

        optimizer.zero_grad()

        outputs = model(

            input_ids=input_ids,

            attention_mask=attention_mask,

            labels=labels

        )

        loss = outputs.loss

        loss.backward()

        optimizer.step()

        scheduler.step()

        train_loss += loss.item()

    train_loss /= len(train_loader)

    ##################################
    # Validation
    ##################################

    model.eval()

    val_loss = 0

    correct = 0

    total = 0

    with torch.no_grad():

        for batch in val_loader:

            input_ids = batch["input_ids"].to(device)

            attention_mask = batch["attention_mask"].to(device)

            labels = batch["labels"].long().to(device)

            outputs = model(

                input_ids=input_ids,

                attention_mask=attention_mask,

                labels=labels

            )

            val_loss += outputs.loss.item()

            predictions = torch.argmax(
                outputs.logits,
                dim=1
            )

            correct += (predictions == labels).sum().item()

            total += labels.size(0)

    val_loss /= len(val_loader)

    accuracy = correct / total

    ##################################
    # W&B Logging
    ##################################

    wandb.log({

        "epoch": epoch + 1,

        "train_loss": train_loss,

        "validation_loss": val_loss,

        "validation_accuracy": accuracy,

        "learning_rate": scheduler.get_last_lr()[0],

        "lowest validation loss": lowest_val_loss

    })

    ##################################
    # Console
    ##################################

    print(

        f"Epoch {epoch+1}/{epochs}"

        f" | Train Loss: {train_loss:.4f}"

        f" | Val Loss: {val_loss:.4f}"

        f" | Accuracy: {accuracy:.4f}"

    )

    ##################################
    # Save Best Model
    ##################################

    if val_loss <= lowest_val_loss:

        lowest_val_loss = val_loss

        model.save_pretrained("best_model")

        tokenizer.save_pretrained("best_model")

        artifact = wandb.Artifact(

            "best-deberta-model",

            type="model"

        )

        artifact.add_dir("best_model")

        wandb.log_artifact(artifact)

print(f"\nBest Validation Accuracy: {accuracy:.4f}")

wandb.finish()
```

```python
import torch
import torch.nn.functional as F

OPTIONS = ["A", "B", "C", "D", "E"]

def predict_mcq(model, tokenizer, row, device):

    model.eval()

    scores = []

    with torch.no_grad():

        for option in OPTIONS:

            encoding = tokenizer(

                row["prompt"],
                row[option],

                truncation=True,
                padding="max_length",
                max_length=256,

                return_tensors="pt"

            )

            input_ids = encoding["input_ids"].to(device)
            attention_mask = encoding["attention_mask"].to(device)

            outputs = model(

                input_ids=input_ids,
                attention_mask=attention_mask

            )

            probabilities = F.softmax(outputs.logits, dim=1)

            score = probabilities[0,1].item()

            scores.append(score)

    ranking = [

        option

        for _, option in sorted(

            zip(scores, OPTIONS),

            reverse=True

        )

    ]

    return ranking, scores
```

```python
predictions = []

for _, row in df_test.iterrows():

    ranking, _ = predict_mcq(

        model,
        tokenizer,
        row,
        device

    )

    predictions.append(

        " ".join(ranking[:3])

    )
```
