## MAP@3 function

```python
# takes two inputs, one is a list of all correct answers, the other a list of corresponding predictions
def map_at_3(y_true, y_pred):
    score = 0.0

    if len(y_true) != len(y_pred):
        raise ValueError("Length mismatch")

    for i in range(len(y_true)):
        pred = y_pred[i]
        true = y_true[i]
        if true in pred:
            for j in range(len(pred)):
                if pred[j] == true:
                    score += 1/(j+1)
                    break
                
    return score/len(y_true)
```

```python
def predict_score(vectorizer, df):
    predictions = []
    
    for _, row in df.iterrows():
    
        prompt_vec = vectorizer.transform(
            [row["prompt"]]
        )
    
        similarities = {}
    
        for option in ["A","B","C","D","E"]:
    
            option_vec = vectorizer.transform(
                [row[option]]
            )
    
            similarities[option] = cosine_similarity(
                prompt_vec,
                option_vec
            )[0,0]
    
        ranked = sorted(
            similarities,
            key=similarities.get,
            reverse=True
        )
    
        predictions.append(
            ranked[:3]
        )
    
    score = map_at_3(
        df["answer"].tolist(),
        predictions
    )
    
    print("Validation MAP@3:", score)

    return score

def tfidf_predict(vectorizer, df):
    predictions = []
    
    for _, row in df.iterrows():
    
        prompt_vec = vectorizer.transform(
            [row["prompt"]]
        )
    
        similarities = {}
    
        for option in ["A","B","C","D","E"]:
    
            option_vec = vectorizer.transform(
                [row[option]]
            )
    
            similarities[option] = cosine_similarity(
                prompt_vec,
                option_vec
            )[0,0]
    
        ranked = sorted(
            similarities,
            key=similarities.get,
            reverse=True
        )
    
        predictions.append(
            ranked[:3]
        )

    return predictions
```

## Word2Vec

```python
# building the corpus
corpus = []

for _, row in train_df.iterrows():
    corpus.append(
        str(clean_text(row['prompt'])).lower().split()
    )

    for option in ["A", "B", "C", "D", "E"]:
        corpus.append(
            str(row[option]).lower().split()
        )
```

```python
# Train word2vec
w2v_model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1,
    seed=42
)
```

```python
def sentence_embedding(text, model):

    words = clean_text(str(text)).lower().split()

    vectors = []

    for word in words:

        if word in model.wv:

            vectors.append(
                model.wv[word]
            )

    if len(vectors) == 0:

        return np.zeros(
            model.vector_size
        )

    return np.mean(
        vectors,
        axis=0
    )
```

### Validation score

```python
predictions = []

for _, row in val_df.iterrows():

    prompt_emb = sentence_embedding(
        clean_text(row["prompt"]),
        w2v_model
    )

    similarities = {}

    for option in ["A","B","C","D","E"]:

        option_emb = sentence_embedding(
            row[option],
            w2v_model
        )

        sim = cosine_similarity(
            prompt_emb.reshape(1,-1),
            option_emb.reshape(1,-1)
        )[0,0]

        similarities[option] = sim

    ranked = sorted(
        similarities,
        key=similarities.get,
        reverse=True
    )

    predictions.append(
        ranked[:3]
    )
```

```python
score = map_at_3(val_df['answer'].tolist(), predictions)
score
```

## sBERT embedding + cosine similarity

```python
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    device="cuda"
)
```

```python
wandb.init(
    entity="24f1002705-IITM",
    project="24f1002705-DLGenAI-t12026",
    name="sbert-baseline",
    config={
        "model": "all-MiniLM-L6-v2"
    }
)

predictions = []

for _, row in df.iterrows():
    prompt_embedding = model.encode(
        row['prompt'],
        convert_to_numpy=True
    )

    similarities = {}

    for option in ["A", "B", "C", "D", "E"]:
        option_embedding = model.encode(
            row[option],
            convert_to_numpy=True
        )
    
        similarity = cosine_similarity(
            prompt_embedding.reshape(1,-1),
            option_embedding.reshape(1, -1)
        )[0][0]
    
        similarities[option] = similarity

    ranked = sorted(
        similarities,
        key=similarities.get,
        reverse=True
    )

    predictions.append(ranked[:3])

score = map_at_3(
    df['answer'].tolist(),
    predictions
)

print(f"score:", score)

wandb.log({
    "map3": score
})

wandb.finish()
```

```python
test_predictions = []

for _, row in df_test.iterrows():

    prompt_embedding = model.encode(
        row["prompt"],
        convert_to_numpy=True
    )

    similarities = {}

    for option in ["A", "B", "C", "D", "E"]:

        option_embedding = model.encode(
            row[option],
            convert_to_numpy=True
        )

        similarities[option] = cosine_similarity(
            prompt_embedding.reshape(1, -1),
            option_embedding.reshape(1, -1)
        )[0, 0]

    ranked = sorted(
        similarities,
        key=similarities.get,
        reverse=True
    )

    test_predictions.append(" ".join(ranked[:3]))
```

## Zero-shot-classification

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device="cuda"
)
```

```python
wandb.init(
    entity="24f1002705-IITM",
    project="24f1002705-DLGenAI-t12026",
    name="zero-shot",
    config={
        "model": "bart-large-mnli"
    }
)

predictions = []

for _, row in df.iterrows():
    labels= [
        row["A"],
        row["B"],
        row["C"],
        row["D"],
        row["E"]
    ]
    
    result = classifier(
        row["prompt"],
        candidate_labels=labels,
        multi_label=False
    )

    ranked_labels = result["labels"][:3]

    ranked_options = []

    for label in ranked_labels:

        if label == row["A"]:
            ranked_options.append("A")

        elif label == row["B"]:
            ranked_options.append("B")

        elif label == row["C"]:
            ranked_options.append("C")

        elif label == row["D"]:
            ranked_options.append("D")

        else:
            ranked_options.append("E")

    predictions.append(ranked_options)

```

```python
score = map_at_3(
    df["answer"].tolist(),
    predictions
)

print(f"Validation MAP@3 : {score:.5f}")

wandb.log({
    "validation_map3": score
})

wandb.finish()
```
